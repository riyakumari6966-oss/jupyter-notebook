```python
def HappyNumber(n):
    def getNext(num):
        total_sum=0
        while num>0:
            digit=num%10
            total_sum+=digit*digit
            num//=10
        return total_sum
    slow=n
    fast=getNext(n)

    while fast!=1 and slow!=fast:
        slow=getNext(slow)
        fast=getNext(getNext(fast))
    return fast==1

n=19
print(HappyNumber(n))
```
