```python
nums=[1,5,4,2,9,9,9]
k=3

if len(nums)<k:
    print(0)

freq={}
curr_sum=0
max_sum=0
for i in range(k):
    curr_sum+=nums[i]
    freq[nums[i]]=freq.get(nums[i],0)+1
if len(freq)==k:
    max_sum=curr_sum
for i in range(k, len(nums)):
    incoming=nums[i]
    outgoing=nums[i-k]

    curr_sum+=incoming
    freq[incoming]=freq.get(incoming,0)+1

    curr_sum-=outgoing
    freq[outgoing]-=1
    if freq[outgoing]==0:
        del freq[outgoing]
    if len(freq)==k:
        max_sum=max(max_sum,curr_sum)
print(max_sum)
```
