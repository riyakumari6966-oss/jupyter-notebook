```python
def MinRecolors(blocks,k):
    min_recolors=float('inf')
    target=['B']*k

    for i in range(len(blocks)-k+1):
        curr_window=list(blocks[i:i+k])
        recolors_count=0

        for j in range(len(curr_window)):
            if curr_window[j]=='W':
                curr_window[j]='B'
                recolors_count+=1

        if curr_window==target:
            min_recolors=min(min_recolors,recolors_count)
    return min_recolors

blocks="WBBWWBBWBW"
k=7
print(MinRecolors(blocks,k))
```
