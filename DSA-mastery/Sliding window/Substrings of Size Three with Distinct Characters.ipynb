```python
s="xyzzaz"
k=3
count=0

for i in range(len(s)-k+1):
    curr_window=list(s[i:i+k])
    has_duplicate=False
    for j in range(len(curr_window)):
        for l in range(j+1,len(curr_window)):
            if curr_window[j]==curr_window[l]:
                has_duplicate=True
                break 

    if not has_duplicate:
        count+=1
print(count)
```
