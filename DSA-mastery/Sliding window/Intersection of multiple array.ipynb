```python
def intersection(nums):
    freq={}
    res=[]

    for i in range(len(nums)):
        for j in range(len(nums[i])):
            freq[nums[i][j]]=freq.get(nums[i][j],0)+1

    for num,count in freq.items():
        if count==len(nums):
            res.append(num)
            res.sort()
    return res

nums=[[3,1,2,4],[1,2,3,4],[6,7,3,4]]
print(intersection(nums))
```
