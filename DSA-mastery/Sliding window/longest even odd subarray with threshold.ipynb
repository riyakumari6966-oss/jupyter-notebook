```python
#brute force approach
def LongestAlternatingSubarray(nums,threshold):
    n=len(nums)
    max_len=0

    for l in range(n):
        if nums[l]%2==0 and nums[l]<=threshold:
            max_len=max(max_len,1)
            for r in range(l+1,n):
                if nums[r]<=threshold and nums[r]%2!=nums[r-1]%2:
                    max_len=max(max_len,r-l+1)
                else:
                    break
    return max_len

nums=[3,2,5,4]
threshold=5
print(LongestAlternatingSubarray(nums,threshold))
```

```python
#sliding window approach
def LongestAlternatingSubarray(nums,threshold):
    n=len(nums)
    l=0
    max_len=0

    while l<n:
        if nums[l]%2==0 and nums[l]<=threshold:
            r=l
            while (r+1<n and nums[r+1]<=threshold and nums[r]%2!=nums[r+1]%2):
                r+=1
            
            max_len=max(max_len,r-l+1)
            l=r+1
        else:
            l+=1
    return max_len

nums=[3,2,5,4]
threshold=5
print(LongestAlternatingSubarray(nums,threshold))
```
