```python
#Maximum number of vowels in a substring of given length 
def max_vowels_inSubstring(s,k):
    list_s=list(s)
    vowels=set("aeiouAEIOU")
    count=0
    result=[]
    for i in range(k):
        if list_s[i] in vowels:
            count+=1
    result.append(count)

    for i in range(k,len(list_s)):
        char_leaving=list_s[i-k]
        if char_leaving in vowels:
            count-=1
        char_entering=list_s[i]
        if char_entering in vowels:
            count+=1
        result.append(count)
    return max(result)
s="Reshika"
k=3
max_vowels_inSubstring(s,k)
```

```python
#Sliding Window approach 
def maxVowelSubstring(s,k):
    if len(s)<k:
        return 0
    s1=s.lower()
    vowel="aeiou"
    curr_vowel=0
    max_vowel=0
    curr_window=s[:k]
    for i in range(len(curr_window)):
        if curr_window[i] in vowel:
            curr_vowel+=1
        max_vowel=max(curr_vowel,max_vowel)
    for i in range(k,len(s1)):
        incoming=s1[i]
        outgoing=s1[i-k]

        if incoming in vowel:
            curr_vowel+=1
        if outgoing in vowel:
            curr_vowel-=1
        max_vowel=max(curr_vowel,max_vowel)
    return max_vowel

s="abcdiiiefg"
k=3
print(maxVowelSubstring(s,k))
    
```
