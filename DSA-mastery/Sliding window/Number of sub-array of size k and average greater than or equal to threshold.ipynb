```python
def numberSubarray(nums,k,threshold):
    if len(nums)<k:
        return 0
    target_sum=k*threshold
    curr_sum=0
    max_sum=0
    count=0
    for i in range(k):
        curr_sum=sum(nums[:k])
        max_sum=max(max_sum,curr_sum)

        if max_sum>=target_sum:
            count+=1
    for i in range(k,len(nums)):
        incoming=nums[i]
        outgoing=nums[i-k]

        curr_sum+=incoming
        curr_sum-=outgoing

        max_sum=max(max_sum,curr_sum)
        if max_sum>=target_sum:
            count+=1
    return count

nums=[2,2,2,2,5,5,5,8]
k=3
threshold=4
print(numberSubarray(nums,k,threshold))
```
