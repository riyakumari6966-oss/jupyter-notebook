```python
#My approach
def maxConsecutiveOnes(nums):
    count=0
    max_count=0
    for i in range(len(nums)):
        if nums[i]==1:
            count+=1
            max_count=max(max_count,count)
        else:
            count=0
    return count

nums=[1,1,0,1,1,1]
print(maxConsecutiveOnes(nums))
```

```python
#Sliding Window approach
def maxConsecutiveOnes(nums):
    left=0
    max_len=0
    for right in range(len(nums)):
        if nums[right]==0:
            left=right+1
        else:
            max_len=max(max_len,right-left+1)
    return max_len

nums=[1,1,0,1,1,1,1]
print(maxConsecutiveOnes(nums))
```
