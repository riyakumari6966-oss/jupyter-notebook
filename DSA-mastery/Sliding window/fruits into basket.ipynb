```python
#brute force
def totalFruits(fruits):
    n=len(fruits)
    max_picked=0

    for i in range(n):
        types=set()

        for j in range(i,n):
            types.add(fruits[j])
            if len(types)>2:
                break
            max_picked=max(max_picked,j-i+1)
    return max_picked

fruits=[0,1,2,1,2]
print(totalFruits(fruits))
```

```python
#optimized approach(Dynamic sliding window (two pointer))
def totalFruits(fruits):
    basket={}
    left=0
    max_picked=0

    for right in range(len(fruits)):
        basket[fruits[right]]=basket.get(fruits[right],0)+1

        while len(basket)>2:
            basket[fruits[left]]-=1
            if basket[fruits[left]]==0:
                del basket[fruits[left]]
            left+=1
        max_picked=max(max_picked,right-left+1)
    return max_picked

fruits=[0,1,2,1,2]
print(totalFruits(fruits))
```
