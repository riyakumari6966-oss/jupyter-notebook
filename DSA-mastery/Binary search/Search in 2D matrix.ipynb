```python
def Findmatrix(matrix, target):
    row=len(matrix)
    cols=len(matrix[0])

    left=0
    right=(row*cols)-1

    while left<=right:
        mid=(left+right)//2
        val=matrix[mid//cols][mid%cols]

        if val==target:
            return True
        elif val<target:
            left=mid+1
        else:
            right=mid-1
    return False

matrix=[[1,3,5,7],[10,11,16,20],[23,30,34,60]]
target=3
print(Findmatrix(matrix,target))
```
