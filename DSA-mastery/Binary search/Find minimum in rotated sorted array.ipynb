```python
def MinimumRotated(nums):
    minimum_val=float('inf')
    left=0
    right=len(nums)-1

    while left<=right:
        mid=(left+right)//2
        minimum_val=min(minimum_val,nums[mid])

        if nums[mid]>nums[right]:
            left=mid+1
        else:
            right=mid-1

    return minimum_val

nums=[4,5,6,7,0,1,2]
print(MinimumRotated(nums))
```
