```python
def sqrt(x):
    if x<2:
        return x
    left=0
    right=x//2
    ans=1

    while left<=right:
        mid=(left+right)//2
        square=mid*mid

        if square==x:
            return mid
        elif square<x:
            ans=mid
            left=mid+1
        else:
            right=mid-1
    return ans

x=4
print(sqrt(x))
```
