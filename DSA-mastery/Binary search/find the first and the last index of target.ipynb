```python
def getFirstLastIndex(nums,target):
    def findBound(isFirst):
        left=0
        right=len(nums)-1
        bound=-1

        while left<=right:
            mid=(left+right)//2

            if nums[mid]==target:
                bound=mid
                if isFirst:
                    right=mid-1
                else:
                    left=mid+1
            elif nums[mid]<target:
                left=mid+1
            else:
                right=mid-1
        return bound

    first=findBound(True)
    if first==-1:
        return [-1,-1]
    last=findBound(False)
    return [first,last]

nums=[5,7,7,8,8,10]
target=8
print(getFirstLastIndex(nums,target))
```
