```python
def minNumberofBouquet(bloomDay,m,k):
    if len(bloomDay)<m*k:
        return -1
    left=min(bloomDay)
    right=max(bloomDay)
    ans=right

    while left<=right:
        mid=(left+right)//2

        consecutive=0
        bouquet=0

        for day in bloomDay:
            if day<=mid:
                consecutive+=1
                if consecutive==k:
                    bouquet+=1
                    consecutive=0
            else:
                consecutive=0
        if bouquet>=m:
            ans=mid
            right=mid-1
        else:
            left=mid+1
    return ans

bloomDay=[1,10,3,10,2]
m=3
k=1
print(minNumberofBouquet(bloomDay,m,k))
```
