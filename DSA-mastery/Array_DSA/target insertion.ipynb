```python
def targetInsertion(nums, target):
    left=0
    right=len(nums)-1

    while left<=right:
        mid=left+(right-left)//2
        if nums[mid]==target:
            return mid
        elif nums[mid]<target:
            left=mid+1
        else:
            right=mid-1
    return left

nums=[1,3,5,6]
target=7
print(targetInsertion(nums, target))
```
