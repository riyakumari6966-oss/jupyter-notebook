```python
def permutationString(s1,s2):
    k=len(s1)
    if k>len(s2):
        return False
    s1_count=[0]*26
    window_count=[0]*26
        
    for i in range(k):
        s1_count[ord(s1[i])-ord('a')]+=1
        window_count[ord(s2[i])-ord('a')]+=1
        
    if s1_count==window_count:
        return True

    for i in range(k,len(s2)):
        window_count[ord(s2[i])-ord('a')]+=1
        window_count[ord(s2[i-k])-ord('a')]-=1

        if s1_count==window_count:
            return True
    return False

s1 = "ab"
s2 = "eidbaooo"
print(permutationString(s1,s2))
```
