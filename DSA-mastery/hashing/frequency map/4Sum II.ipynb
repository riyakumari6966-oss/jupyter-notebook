```python
def FourSumCount(nums1,nums2,nums3,nums4):
    sum_ab={}
    count=0

    for a in nums1:
        for b in nums2:
            total=a+b
            sum_ab[total]=sum_ab.get(total,0)+1
    for c in nums3:
        for d in nums4:
            target=-(c+d)
            count+=sum_ab.get(target,0)
    return count

nums1=[1,2]
nums2=[-2,-1]
nums3=[-1,2]
nums4=[0,2]
print(FourSumCount(nums1,nums2,nums3,nums4))
```
