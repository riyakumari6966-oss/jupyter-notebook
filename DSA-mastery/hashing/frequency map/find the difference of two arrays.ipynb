```python
def findDifference(nums1,nums2):
    freq_nums1={}
    freq_nums2={}

    for num in nums1:
        freq_nums1[num]=freq_nums1.get(num,0)+1
    for num in nums2:
        freq_nums2[num]=freq_nums2.get(num,0)+1
    
    list1=[]
    for key in freq_nums1:
        if key not in freq_nums2:
            list1.append(key)

    list2=[]
    for key in freq_nums2:
        if key not in freq_nums1:
            list2.append(key)

    return [list1, list2]

nums1=[1,2,3]
nums2=[2,4,5]
print(findDifference(nums1,nums2))
```

```python
def findDifference(nums1,nums2):
    s1, s2=set(nums1),set(nums2)
    return[list(s1-s2), list(s2-s1)]

nums1=[1,2,3]
nums2=[2,4,5]
print(findDifference(nums1,nums2))
```
