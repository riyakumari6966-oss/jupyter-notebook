```python
def validAnagram(s,t):
    freq_s={}
    freq_t={}

    for i in range(len(s)):
        freq_s[s[i]]=freq_s.get(s[i],0)+1
    for i in range(len(t)):
        freq_t[t[i]]=freq_t.get(t[i],0)+1
    if freq_s==freq_t:
        return True
    return False

s="anagram"
t="nagaram"
print(validAnagram(s,t))
```
