```python
def wordPattern(pattern,s):
    pattern_to_s={}
    s_to_pattern={}
    words=s.split()
    if len(words)!=len(pattern):
        return False

    for ch1, word in zip(pattern, words):
        if ch1 in pattern_to_s and pattern_to_s[ch1]!=word:
            return False
        if word in s_to_pattern and s_to_pattern[word]!=ch1:
            return False

        pattern_to_s[ch1]=word
        s_to_pattern[word]=ch1
    return True

pattern='abba'
s="dog cat cat dog"
print(wordPattern(pattern,s))
```
