```python
def canConstruct(ransomNote,magazine):
    freq_ransom={}
    freq_magazine={}

    for i in ransomNote:
        freq_ransom[i]=freq_ransom.get(i,0)+1
    for i in magazine:
        freq_magazine[i]=freq_magazine.get(i,0)+1
    for key in freq_ransom:
        if freq_magazine.get(key,0)<freq_ransom[key]:
            return False
    return True
ransomNote="aa"
magazine="aab"
print(canConstruct(ransomNote,magazine))
```
