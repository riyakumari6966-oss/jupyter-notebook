```python
def NextGreaterEle(nums1,nums2):
    stack=[]
    greater={}

    for num in nums2:
        while stack and num>stack[-1]:
            smaller_number=stack.pop()
            greater[smaller_number]=num
        stack.append(num)
    return [greater.get(num,-1) for num in nums1]

nums1=[4,1,2]
nums2=[1,3,4,2]
print(NextGreaterEle(nums1,nums2))
```
