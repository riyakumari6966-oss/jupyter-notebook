```python
def finalPrice(prices):
    answer=prices[:]
    stack=[]

    for i,current_price in enumerate(prices):
        while stack and current_price<=prices[stack[-1]]:
            idx=stack.pop()
            answer[idx]-=current_price
        stack.append(i)
    return answer

prices=[8,4,6,2,3]
print(finalPrice(prices))
```
