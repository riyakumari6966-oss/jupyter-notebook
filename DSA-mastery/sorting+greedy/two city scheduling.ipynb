```python
def Cityschedule(costs):
    costs.sort(key=lambda x:x[0]-x[1])

    total=0
    half=len(costs)//2

    for i in range(len(costs)):
        if i<half:
            total+=costs[i][0]
        else:
            total+=costs[i][1]
    return total

costs=[[259,770],[448,54],[926,667],[184,139],[840,118],[577,469]]
print(Cityschedule(costs))
```
