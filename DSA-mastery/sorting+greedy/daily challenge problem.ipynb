```python
#Approach 1 O(nlogn)
def targetIndices(nums, target):
    nums.sort()
    res=[]
    for i in range(len(nums)):
        if nums[i]==target:
            res.append(i)
    return res
nums=[1,2,5,2,3]
target=2
print(targetIndices(nums,target))
```

```python
#Approach 2 O(n)
def targetIndices(nums, target):
    less_count=0
    target_count=0
    for num in nums:
        if num<target:
            less_count+=1
        elif num==target:
            target_count+=1
    return list(range(less_count,less_count+target_count))

nums=[1,2,5,2,3]
target=2
print(targetIndices(nums,target))
```
