```python
def SubarraySum(nums,k):
    count=0
    current_sum=0
    prefix_map={0:1}

    for num in nums:
        current_sum+=num
        complement=current_sum-k

        if complement in prefix_map:
            count+=prefix_map[complement]
        prefix_map[current_sum]=prefix_map.get(current_sum,0)+1
    return count

nums=[1,2,1,2,1]
k=3
print(SubarraySum(nums,k))
```
