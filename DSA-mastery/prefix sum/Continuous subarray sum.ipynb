```python
def ContinuousSum(nums,k):
    remainder_map={0:-1}
    current_sum=0

    for j,num in enumerate(nums):
        current_sum+=num
        remainder=current_sum % k

        if remainder in remainder_map:
            if j-remainder_map[remainder]>=2:
                return True
        else:
            remainder_map[remainder]=j
    return False

nums=[23, 2, 4, 6, 7]
k=6
print(ContinuousSum(nums, k))
```
