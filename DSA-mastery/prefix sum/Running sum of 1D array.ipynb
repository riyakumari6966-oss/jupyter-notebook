```python
def RunningSum(nums):
    res=[]
    res.append(nums[0])
    for i in range(1,len(nums)):
        res.insert(i,sum(nums[:i+1]))
    return res

nums=[1,2,3,4]
print(RunningSum(nums))
```

```python
def RunningSum(nums):
    for i in range(1,len(nums)):
        nums[i]+=nums[i-1]
    return nums

nums=[1,2,3,4]
print(RunningSum(nums))
```
