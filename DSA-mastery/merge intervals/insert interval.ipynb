```python
def insert_and_merge(intervals,new_interval):
    intervals.append(new_interval)
    intervals.sort(key=lambda x:x[0])
    merged=[intervals[0]]
    for current in intervals[1:]:
        last_merged=merged[-1]
        if current[0]<=last_merged[1]:
            last_merged[1]=max(last_merged[1], current[1])
        else:
            merged.append(current)
    return merged

if __name__=="__main__":
    intervals=[[1,3],[6,9]]
    new_interval=[2,5]
    print(insert_and_merge(intervals,new_interval))
```

```python
#take O(nlogn) time complexity because of sorting the intervals.
def insertInterval(intervals,newIntervals):
    intervals.append(newIntervals)
    intervals.sort(key=lambda x:x[0])
    merged=[intervals[0]]
    for current in intervals[1:]:
        last_merged=merged[-1]
        if current[0]<=last_merged[1]:
            last_merged[1]=max(last_merged[1],current[1])
        else:
            merged.append(current)
    return merged
```

```python
#O(n) time complexity solution without sorting the intervals.
def insertIntervalOptimized(intervals,newIntervals):
    merged=[]
    i=0
    while i<len(intervals) and intervals[i][1]<newIntervals[0]:
        merged.append(intervals[i])
        i+=1
    while i<len(intervals) and intervals[i][0]<=newIntervals[1]:
        newIntervals[0]=min(newIntervals[0],intervals[i][0])
        newIntervals[1]=max(newIntervals[1],intervals[i][1])
        i+=1
    merged.append(newIntervals)

    while i<len(intervals):
        merged.append(intervals[i])
        i+=1
    return merged
```
