```python
def poisonInterval(timeSeries, duration):
    if not timeSeries or duration==0:
        return 0
    total_time=0
    for i in range(len(timeSeries)-1):
        gap=timeSeries[i+1]-timeSeries[i]
        total_time+=min(gap,duration)
    total_time+=duration
    return total_time

timeSeries = [1, 4]
duration = 2
print(poisonInterval(timeSeries, duration))  
```
