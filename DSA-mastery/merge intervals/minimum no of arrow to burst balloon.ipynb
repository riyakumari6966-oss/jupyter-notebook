```python
def findMinArrowShots(points):
    if not points:
        return 0
    points.sort(key=lambda x:x[1])
    arrow=1
    prev_end=points[0][1]

    for start,end in points[1:]:
        if start>prev_end:
            arrow+=1
            prev_end=end
    return arrow

points=[[10,16],[2,8],[1,6],[7,12]]
print(findMinArrowShots(points))
```
