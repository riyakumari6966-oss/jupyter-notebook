```python
def removeInterval(intervals):
    if not intervals:
        return []

    intervals.sort(key=lambda x:(x[0],-x[1]))

    remaining=0
    prev_end=0

    for start, end in intervals:
        if start<prev_end:
            remaining+=1
        else:
            prev_end=end
    return remaining

intervals = [[1,3],[2,4],[3,5],[6,8]]
print(removeInterval(intervals))  
```
