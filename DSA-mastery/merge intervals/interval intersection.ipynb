```python
#brute force approach
def interval_intersection(first,second):
    intersection=[0]*2
    if second[0]<=first[1] and second[1]>=first[1]:
        intersection[0]=second[0]
        intersection[1]=first[1]
        return intersection
    elif second[0]<=first[0] and second[1]<=first[1]:
        intersection[0]=first[0]
        intersection[1]=second[1]
        return intersection
    elif second[0]>=first[0] and second[1]<=first[1]:
        intersection[0]=second[0]
        intersection[1]=second[1]
        return intersection
    else:
        return []
first=[1,5]
second=[2,7]
print(interval_intersection(first,second))
first1=[2,7]
second1=[1,5]
print(interval_intersection(first1,second1))
first2=[1,3]
second2=[6,9]
print(interval_intersection(first2,second2))
first3=[1,9]
second3=[5,8]
print(interval_intersection(first3,second3))
```

```python
#optimized way
def interval_intersection(first,second):
    start=max(first[0],second[0])
    end=min(first[1],second[1])
    if start<=end:
        return[start,end]
    else:
        return []
first=[1,5]
second=[2,7]
print(interval_intersection(first,second))
first1=[2,7]
second1=[1,5]
print(interval_intersection(first1,second1))
first2=[1,3]
second2=[6,9]
print(interval_intersection(first2,second2))
first3=[1,9]
second3=[5,8]
print(interval_intersection(first3,second3))
```
