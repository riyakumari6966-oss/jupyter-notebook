```python
def intersectpoint(intervals):
    if not intervals:
        return None
    intervals.sort(key=lambda x:x[0])
    merged=[intervals[0]]

    for start,end in intervals[1:]:
        if start<=merged[-1][1]:
            merged[-1][1]=max(merged[-1][1],end)
        else:
            merged.append([start,end])
    count=0
    for start,end in merged:
        count+=end-start+1
    return count

intervals=[[1,3],[2,4],[5,7]]
print(intersectpoint(intervals))
```
