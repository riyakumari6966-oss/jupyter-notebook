```python
def missing_ranges(nums,lower,upper):
    result=[]
    if not nums:
        return[[lower,upper]]
    
    if nums[0]>lower:
        result.append([lower,nums[0]-1])

    for i in range(1,len(nums)):
        if nums[i-1]!=nums[i]-1:
            result.append([nums[i-1]+1,nums[i]-1])
            
    if nums[-1]<upper:
        result.append([nums[len(nums)-1]+1,upper])
    return result

nums=[1,3,50,75]
lower=0
upper=99
print(missing_ranges(nums,lower,upper))
nums=[0,1,3,50,75]
lower=0
upper=99
print(missing_ranges(nums,lower,upper))
nums=[1,3,50,75,99]
lower=0
upper=99
print(missing_ranges(nums,lower,upper))


```
